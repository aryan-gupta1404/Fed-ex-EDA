# **Project Summary -**

This project is basically about trying to understand how FedEx deliveries are working in real life using data. I worked with a dataset that had information about shipments like which country the delivery was going to, which vendor handled it, how it was shipped (air, truck, etc.), delivery dates, and costs.

The main idea was to figure out why some deliveries are getting delayed and what factors are causing those delays. To make this easier, I created a new column called delay_days, which simply shows the difference between when a delivery was supposed to arrive and when it actually arrived. This helped me clearly see which deliveries were late, early, or on time.

From there, I started breaking things down. First, I looked at different countries to see where delays were happening the most. Then I went one step deeper and checked which vendors in those countries were causing the issues. I also looked at shipment modes like air and truck to understand how the method of delivery affects time and cost.

I also explored things like how many deliveries each country is handling and whether spending more on shipping actually reduces delays. To make all this easier to understand, I used simple graphs like bar charts and scatter plots.

#### Chart - 5

```python
# Chart - 5 visualization code

congo_clean = congo_df[congo_df['Vendor'] != 'SCMS from RDC']
before = congo_df['delay_days'].mean()
after = congo_clean['delay_days'].mean()
labels = ['Before', 'After']
values = [before, after]

plt.figure(figsize=(6,4))
sns.barplot(x=labels, y=values)
plt.ylabel('Average Delay (days)')
plt.title('Impact of Removing Problematic Vendor in Congo')
plt.show()
```

##### 1. Why did you pick the specific chart?

This bar chart was specifically chosen to visually compare the average delay days in Congo, DRC 'Before' and 'After' the hypothetical removal of the problematic 'SCMS from RDC' vendor. It's an effective way to demonstrate the direct impact of a strategic decision by comparing two distinct states of a numerical variable.

##### 2. What is/are the insight(s) found from the chart?

Insights:
1- The most striking insight is the dramatic reduction in average delay days for Congo, DRC after excluding 'SCMS from RDC' from the dataset. The average delay drops from over 11 days to approximately 1-2 days.
2- This significant improvement unequivocally confirms that 'SCMS from RDC' was the major bottleneck and primary cause of delivery inefficiencies in the region.
3- The comparison highlights the profound positive effect that vendor optimization can have on logistics performance.

##### 3. Will the gained insights help creating a positive business impact?

2- **Cost Savings**: Reduced delays can lead to fewer penalties, less need for expedited shipping to compensate for delays, and better inventory management, resulting in significant cost savings.

#### Chart - 10

```python
# Chart - 10 visualization code
vendor_analysis = df.groupby('Vendor').agg({
    'delay_days': 'mean',
    'Freight Cost (USD)': 'mean'
}).reset_index()

vendor_analysis = vendor_analysis.sort_values(by='delay_days', ascending=False).head(10)

sns.scatterplot(
    x='Freight Cost (USD)',
    y='delay_days',
    data=vendor_analysis)

for i in range(len(vendor_analysis)):
    plt.text(
        vendor_analysis['Freight Cost (USD)'].iloc[i],
        vendor_analysis['delay_days'].iloc[i],
        vendor_analysis['Vendor'].iloc[i],
        fontsize=8
    )
```

I chose a scatter plot to visualize the relationship between 'Freight Cost (USD)' and 'delay_days' for different vendors. This type of chart is effective for identifying clusters, outliers, and potential correlations between two numerical variables, while also allowing for the labeling of individual data points (vendors).

##### 2. What is/are the insight(s) found from the chart?

Insights:
1- The scatter plot reveals that some vendors contribute to high average delays, irrespective of their freight costs. For instance, 'SCMS from RDC' consistently shows high delays.
2- There isn't a direct linear correlation between freight cost and delay days for all vendors; some vendors might be expensive but still cause delays, while others might be cheaper and also cause delays.
3- Identifying vendors that offer both high freight costs and high delay days is crucial for cost optimization and efficiency improvements.
